1. wap to find the mean of the data items without using function

```python
li = [34,55,22,55,67,78]
sum = 0
for i in li:
  sum = sum + i
mean = sum/len(li)
print(f"The mean of the data points is {mean}")
```

2.WAP to find the mean with using function

```python
import numpy as np
li = []
n = int(input("enter the no of elements in the list"))
for i in range(n):
  ele = int(input("enter the elements:"))
  li.append(ele)
  mean = np.mean(li)
print("mean of the element is",mean)
```

```python
li = []
n = int(input("enter the number of elements in the list"))

for i in range(n):
  ele = int(input("enter the elemrnts:"))
  li.append(ele)

li.sort()
print(li)

if n % 2 == 0:
    med1 = li[n//2]
    med2 = li[n//2 -1]
    median = (med1+ med2)/2
else:
    median= li[n//2]
print("median is",median)
```
