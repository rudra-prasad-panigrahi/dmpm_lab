```python
# classification using decision tree induction(ID3)
import numpy as np
import pandas as pd

df = pd.read_csv('weather.csv')
print(df)
```

```python
count_yes = 0
for i in df['play']:
  if i == 'yes':
    count_yes += 1
count_no = len(df['play']) - count_yes
print("yes=",count_yes)
print("no=",count_no)
```

```python
#total entropy
total_entropy = - (count_yes/len(df['play']))*np.log2(count_yes/len(df['play'])) - (count_no/len(df['play']))*np.log2(count_no/len(df['play']))
print("total_entropy=",total_entropy)
```

```python
#outlook
outlook_unique = df['outlook'].unique()
print(outlook_unique)

```

```python
#calculating each outlook
for i in outlook_unique:
  count_yes = 0
  count_no = 0
  for j in range(len(df['outlook'])):
    if df['outlook'][j] == i and df['play'][j] == 'yes':
      count_yes += 1
    elif df['outlook'][j] == i and df['play'][j] == 'no':
      count_no += 1
  print(f"Outlook: {i}")
  print(f"  Count 'yes': {count_yes}")
  print(f"  Count 'no': {count_no}")
  print(f"  Total instances: {count_yes + count_no}")
  print("entropy(outlook)",- (count_yes/(count_yes + count_no))*np.log2(count_yes/(count_yes + count_no)) - (count_no/(count_yes + count_no))*np.log2(count_no/(count_yes + count_no)))
  print("--------------------")
```

```python
# Calculate weighted entropy for 'outlook'
weighted_entropy_outlook = 0

for i in outlook_unique:
  count_yes = 0
  count_no = 0
  total_instances_outlook = 0
  for j in range(len(df['outlook'])):
    if df['outlook'][j] == i:
      total_instances_outlook += 1
      if df['play'][j] == 'yes':
        count_yes += 1
      elif df['play'][j] == 'no':
        count_no += 1

  if count_yes == 0 or count_no == 0:
    entropy_outlook_i = 0
  else:
    prob_yes = count_yes / total_instances_outlook
    prob_no = count_no / total_instances_outlook
    entropy_outlook_i = -prob_yes * np.log2(prob_yes) - prob_no * np.log2(prob_no)

  # Add weighted entropy for this outlook category
  weighted_entropy_outlook += (total_instances_outlook / len(df['play'])) * entropy_outlook_i

print(f"Weighted Entropy for Outlook: {weighted_entropy_outlook}")
```

```python
information_gain = total_entropy - weighted_entropy_outlook
print("information_gain=",information_gain)
```
