Factorial of a number using function

```python
def factorial(n):
  if n < 0:
    return "factorial is not defined for negative numbers"
  elif n == 0:
    return 1
  else:
    fact = 1
    for i in range(1, n + 1):
      fact *= i
    return fact

num = 5
result = factorial(num)
print(f"The factorial of {num} is {result}")
```

count the frequence of given elements in a list

```python
li = ['a', 's', 'b', 'a', 'm', 'k']
frequency = {}
for item in li:
  frequency[item] = frequency.get(item, 0) + 1

print("Element frequencies:", frequency)
```

find maximum and minimum values in a list

```python
li = [2,3,390,4,44,78]
max = li[0]
min = li[0]
for i in li:
  if i > max:
    max = i
  elif i < min:
    min = i
print(f"The maximum value is {max}")
print(f"The minimum value is {min}")
```
