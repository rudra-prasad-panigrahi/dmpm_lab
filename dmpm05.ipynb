chi-square Test

```python
import numpy as np
import pandas as pd

data = {
    'flavour' : ['strawberry','coffee', 'orange', 'vanilla'],
    'Male' : [23, 18, 8, 8],
    'Female' : [15, 6, 12, 10]
}
df = pd.DataFrame(data)
df
```

```python
#for expected table
# E = (row total * col total)/ grand total
```

```python
df_transposed = df.set_index('flavour')[['Male', 'Female']].T
display(df_transposed)
```

```python
h0 = 'flavour preference is independent of gender'
h1 = 'flavour preference is dependent of gender'
print(h0)
print(h1)
```

```python
degrees_of_freedom = (df_transposed.shape[0] - 1) * (df_transposed.shape[1] - 1)
print(degrees_of_freedom)
```

```python
row_totals = df_transposed.sum(axis=1)
col_totals = df_transposed.sum(axis=0)
grand_total = df_transposed.values.sum()

display(row_totals)
display(col_totals)
print(grand_total)
```

```python
expected_row_totals = col_totals / grand_total
expected_col_totals = row_totals / grand_total


display(expected_row_totals)
display(expected_col_totals)
```
